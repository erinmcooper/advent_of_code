```python
def part1(filename):
    with open(filename, "r") as f:
        lines = f.readlines()
    
    start = 50
    current = start
    zero_count = 0
    for line in lines:
        direction = line[0]
        value = int(line[1:])
        if direction == "L":
            new = current - value
            current = new % 100
            if current == 0:
                zero_count+=1
        if direction == "R":
            new = current + value
            current = new % 100
            if current == 0:
                zero_count+=1
    print(f"PASSWORD: {zero_count}")
```

```python
part1("test.txt")
```

```python
part1("input1.txt")
```

```python
part1("reddit.txt")
```

```python
def part2(filename, debug=False):
    with open(filename, "r") as f:
        lines = f.readlines()
    
    start = 50
    current = start
    zero_count = 0
    zero_crossings = 0
    changed_sign = 0
    for line in lines:
        direction = line[0]
        value = int(line[1:])
        additional, value = divmod(value, 100)
        zero_crossings+=additional
        old = current
        if direction == "L":
            new = current - value
            
        if direction == "R":
            new = current + value
        
        additional, current = divmod(new,100)
        if current == 0:
            zero_count+=1
        if new > 100:
            zero_crossings+= additional
        if (new < 0) and (old > 0): # starts at zero and goes left, don't over count
            changed_sign+=1
        if new < -100:
            zero_crossings+=abs(additional)-1
        if debug:
            print(f"{direction} {value=:4d} {old=:3d} {new=:4d} {current=:3d} {zero_count=:2d} {changed_sign} {zero_crossings=} PASSWORD {zero_count+changed_sign+zero_crossings}")
    print(f"{zero_count=:4d} {changed_sign=:4d} {zero_crossings=:4d} PASSWORD: {zero_count+changed_sign+zero_crossings}")
```

```python
part2("test.txt", debug=True)
```

```python
part2("input1.txt")
```

```python
part2("reddit.txt", debug=True)
```

```python
part2("fake.txt", debug=True)
```

```python
part2("reddit2.txt", debug=True)
```
