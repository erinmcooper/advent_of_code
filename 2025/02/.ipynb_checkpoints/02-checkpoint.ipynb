```python
def read_input(filename):
    with open(filename, "r") as f:
        data = f.readlines()
    data = data[0].split(",")
    return data
```

```python
def is_number_mirror(num):
    
    num_str = str(num)
    if len(num_str)%2 != 0:
        return False
    
    halflength = int(len(num_str)/2)
    if num_str[:halflength]==num_str[halflength:]:
        return True
    else:
        return False
```

```python
def part1(filename, debug=False):
    data = read_input(filename)
    
    invalid_sum = 0
    for id_range in data:
        start, end = [int(val) for val in id_range.split("-")]
        if debug:
            print(f"{id_range=} {start=} {end=}")
        for num in range(start,end+1):
            if is_number_mirror(num):
                if debug:
                    print(f"invalid={num}")
                invalid_sum+=num
    print(f"{invalid_sum=}")
```

```python
part1("test.txt")
```

```python
part1("input1.txt")
```

```python
import math

def is_number_duplicated(num, debug=False):
    
    num_str = str(num)
    length = len(num_str)
    halflength = length/2
    start = 1
    if isinstance(halflength, int):
        stop=math.ceil(halflength)
    else:
        stop=math.floor(halflength)
    #print(f"{start=} {stop=}")
    for i in range(start,stop+1):
        quotient, remainder = divmod(length, i)
        #print(f"{i=} {quotient=} {remainder=}")
        if remainder != 0:
            continue
        else:
            check = [num_str[i*j:i*(j+1)] for j in range(quotient)]
            #print(f"{i=} {quotient=} {check=}")
            if len(set(check)) == 1:
                if debug:
                    print(f"{i=} {quotient=} {check=}")
                return True
```

```python
def part2(filename, debug=False):
    data = read_input(filename)
    
    invalid_sum = 0
    for id_range in data:
        start, end = [int(val) for val in id_range.split("-")]
        if debug:
            print(f"{id_range=} {start=} {end=}")
        for num in range(start,end+1):
            if is_number_duplicated(num):
                if debug:
                    print(f"  INVALID {num}")
                invalid_sum+=num
    #print(f"{invalid_sum=}")
    return invalid_sum
```

```python
assert part2("test.txt")==4174379265
```
