```python
def find_bank_max_joltage_2(bank, debug=False):
    
    maxpair = int(bank[:2])
    for i in range(len(bank)):
        battery1 = bank[i]
        for j in range(i+1,len(bank)):
            battery2 = bank[j]
            testpair = int(battery1+battery2)
            if testpair>maxpair:
                maxpair=testpair
            #if debug:
            #    print(f"{bank=} {maxpair=} {i=} {j=} {battery1=} {battery2=}")
    if debug:
        print(f"{bank=} {maxpair=}")
    return maxpair
```

```python
def part1(filename, debug=False):
    
    with open(filename, "r") as f:
        lines = f.readlines()
    lines = [line.split("\n")[0] for line in lines]
    
    max_joltage = 0
    for bank in lines:
        max_joltage += find_bank_max_joltage_2(bank, debug=debug)
        
    print(f"{max_joltage=}")
    
    return max_joltage
```

```python
assert part1("test.txt", debug=True) == 357
```

```python
def find_next_index(curr, value, bank):
    for idx, char in enumerate(bank):
        if (idx > curr) and (char==value):
            return idx

def find_bank_max_joltage_num_batts(bank, num_batts=12, debug=False):
    
    banklist = list(bank)
    length = len(bank)
    search = banklist[:-num_batts+1]
    if debug:
            print(f"      {''.join(search)}  {max(search)}")
    maxbattarray = [banklist.index(max(search))]
    remaining = num_batts-1
    while len(maxbattarray) != num_batts:
        curr = maxbattarray[-1]
        if remaining != 1:
            search = bank[curr+1:-(remaining-1)]
        else:
            search = bank[curr+1:]

        value = max(list(search))
        if debug:
            #print(f"{remaining=} {curr=} {search=} {''.join([bank[i] for i in maxbattarray])}")
            print(f"      {''.join(search)} {max(search)}")
        maxbattarray.append(find_next_index(curr, value, bank))
        remaining -= 1
    
    maxbatts = int("".join([bank[i] for i in maxbattarray]))
    if debug:
        print(f"      {''.join([bank[i] if (i in maxbattarray) else '.' for i in range(len(bank))])}")
        print(f"{bank=}")
    
    return maxbatts
```

```python
def part2(filename, num_batts=12, debug=False):
    
    with open(filename, "r") as f:
        lines = f.readlines()
    lines = [line.split("\n")[0] for line in lines]
    
    max_joltage_array = []
    for bank in lines:
        max_joltage_array.append(find_bank_max_joltage_num_batts(bank, num_batts=num_batts, debug=debug))
        #if debug:
        #   print(f"{bank=} {max_joltage_array[-1]}")
    max_joltage = sum(max_joltage_array)
    print(f"{max_joltage=}")
    
    return max_joltage
```
