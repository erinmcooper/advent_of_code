```python
import numpy as np
```

```python
def read_input(filename):
    with open(filename, "r") as f:
        data = f.readlines()
    data = [list(line.split("\n")[0]) for line in data]
    return data
```

```python
i = 0
j = 0
for i in range(len(data)):
    for j in range(len(data[i])):
        if data[i][j] == "@":
            adjacent = []
            left = i-1
            right = i+1
            up = j-1
            down = j+1

            if (left > 0) and (up > 0):
                if data[left][up] in ["@","x"]:
                    adjacent.append(data[left][up])
            if up > 0:
                if data[i][up] in ["@","x"]:
                    adjacent.append(data[i][up])
            if (right < len(data[i])) and (up > 0):
                if data[right][up] in ["@","x"]:
                    adjacent.append(data[right][up])

            if left > 0:
                if data[left][j] in ["@","x"]:
                    adjacent.append(data[left][j])
            if right < len(data[i]):
                if data[right][j] in ["@","x"]:
                    adjacent.append(data[right][j])


            if (left > 0) and (down < len(data[j])):
                if data[left][down] in ["@","x"]:
                    adjacent.append(data[left][down])
            if down < len(data[j]):
                if data[i][down] in ["@","x"]:
                    adjacent.append(data[i][down])
            if (right < len(data[i])) and (down < len(data[j])):
                if data[right][down] in ["@","x"]:
                    adjacent.append(data[right][down])

            if adjacent.count("@")+adjacent.count("x")<4:
                data[i][j] = "x"
```
