```python
def read_input(filename):
    with open(filename, "r") as f:
        data = f.readlines()
    data = [list(line.split("\n")[0]) for line in data]
    return data
```

```python
def mark_xs(data, debug=False):
    i = 0
    j = 0
    for i in range(len(data)):
        for j in range(len(data[i])):
            if data[i][j] == "@":
                adjacent = {}
                left = i-1
                right = i+1
                up = j-1
                down = j+1

                if (left >= 0) and (up >= 0):
                    if data[left][up] in ["@","x"]:
                        adjacent["lu"] = data[left][up]
                if up >= 0:
                    if data[i][up] in ["@","x"]:
                        adjacent["u"] = data[i][up]
                if (right < len(data[i])) and (up >= 0):
                    if data[right][up] in ["@","x"]:
                        adjacent["ru"] = data[right][up]

                if left >= 0:
                    if data[left][j] in ["@","x"]:
                        adjacent["l"] = data[left][j]
                if right < len(data[i]):
                    if data[right][j] in ["@","x"]:
                        adjacent["r"] = data[right][j]


                if (left >= 0) and (down < len(data[j])):
                    if data[left][down] in ["@","x"]:
                        adjacent["ld"] = data[left][down]
                if down < len(data[j]):
                    if data[i][down] in ["@","x"]:
                        adjacent["d"] = data[i][down]
                if (right < len(data[i])) and (down < len(data[j])):
                    if data[right][down] in ["@","x"]:
                        adjacent["rd"] = data[right][down]
                #if list(adjacent.values()).count("@")+list(adjacent.values()).count("x")<4:
                if len(adjacent) < 4:
                    data[i][j] = "x"
                    if debug:
                        print(i,j,len(data[i]), len(data[j]), left, right, up, down, data[i][j],adjacent)
    return sum([i.count("x") for i in data]), data
```

```python
def remove_rolls(data):
    
    data = [
        list(map(lambda x: "." if x=="x" else x, i))
        for i in data
    ]
    return data
```

```python
def print_data(data):
    for i in data:
        print("".join(i))
```

```python
def part2(filename, debug=False):
    
    data = read_input(filename)
    accessible_rolls, data = mark_xs(data)
    removable_rolls = accessible_rolls
    test = sum(["x" in i for i in data])>0
    if debug:
        print_data(data)
        print(accessible_rolls)
    while test:
        data = remove_rolls(data)
        accessible_rolls, data = mark_xs(data)
        removable_rolls+=accessible_rolls
        test = sum(["x" in i for i in data])>0
        if debug:
            print_data(data)
            print(accessible_rolls)
    print(f"{removable_rolls=}")
    return removable_rolls
```

```python
part2("test.txt", debug=True)
```

```python
assert part2("test.txt") == 43
```
