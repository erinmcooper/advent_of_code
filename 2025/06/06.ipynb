```python
import math
```

```python
def read_input(filename):
    with open(filename, "r") as f:
        data = f.readlines()
    operations = data.pop(-1).split()
    data = [line.split("\n")[0].split() for line in data]
    return data, operations
```

```python
data, operations = read_input("test.txt")
```

```python
operations
```

```python
translate_operations = {
    "*": math.prod,
    "+": sum,
}
```

```python
def part1(filename, debug=False):
    data, operations = read_input(filename)
    test_sum = 0
    for operation, i in zip(operations, range(len(data[0]))):
        nums_to_operate = [int(data[j][i]) for j in range(len(data))]
        operation_result = translate_operations[operation](nums_to_operate)
        print(operation, nums_to_operate, operation_result)
        test_sum+=operation_result
    print(test_sum)
    return test_sum
```

```python
assert part1("test.txt") == 4277556
```

```python
def read_vertical_input(filename):
    with open(filename, "r") as f:
        data = f.readlines()
    operations = data.pop(-1).split()
    data = [list(line.split("\n")[0]) for line in data]
    
    pivoted = list(map(list, zip(*data)))
    pivoted_list = ["".join(line) for line in pivoted]
    
    groups = []
    current = []
    for item in pivoted_list:
        if item.strip() == "":
            if current:
                groups.append(current)
                current = []
        else:
            current.append(int(item))
    groups.append(current)
    
    return groups, operations
```

```python
data, operations = read_vertical_input("test.txt")
```

```python
for line,operation in zip(data, operations):
    print(line, operation)
```

```python
def part2(filename, debug=False):
    data, operations = read_vertical_input(filename)
    test_sum = 0
    for operation, nums_to_operate in zip(operations, data):
        operation_result = translate_operations[operation](nums_to_operate)
        print(operation, nums_to_operate, operation_result)
        test_sum+=operation_result
    print(test_sum)
    return test_sum
```

```python
assert part2("test.txt") == 3263827
```
