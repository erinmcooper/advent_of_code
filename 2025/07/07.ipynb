```python
def read_input(filename):
    with open(filename, "r") as f:
        data = f.readlines()
    
    for i in range(len(data)):
        data[i] = data[i].split("\n")[0]
        data[i] = list(data[i])
    # turn the S into the incoming beam
    for i in range(len(data[0])):
        if data[0][i]=="S":
            data[0][i]="|"
    return data
```

```python
def split_the_beams(data, debug=False):
    
    beams_to_pass = []
    splitters = []
    for idx, line in enumerate(data):
        if beams_to_pass:
            for beam in beams_to_pass:
                if data[idx][beam] == ".":
                    data[idx][beam] = "|"
        splitters = [i for i,v in enumerate(data[idx]) if v=="^"]
        if splitters:
            for splitter in splitters:
                if splitter-1>=0:
                    if data[idx][splitter-1]==".":
                        data[idx][splitter-1]="|"
                if beam+1<=len(data[idx])-1:
                    if data[idx][splitter+1]==".":
                        data[idx][splitter+1]="|"
        beams_to_pass = [i for i,v in enumerate(data[idx]) if v=="|"]
        if debug:
            print(f"{idx:02d}", "".join(line))#, beams_to_pass, splitters)
    return data
```
