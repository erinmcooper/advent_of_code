```python
def read_input(filename):
    data = []
    with open(filename, "r") as f:
        for line in f.readlines():
            data.append(tuple([int(thing) for thing in line.split(",")]))
        
    return data
```

```python
data_input = read_input("test.txt")
```

```python
data
```

```python
import math
```

```python
data = {}
for point in data_input:
    
    temp = {}
    for i in data_input:
        if i!=point:
            temp[i] = math.dist(point,i)
    nearest_point = min(temp, key=temp.get)
    data[point] = {"nearest": nearest_point, "distance": math.dist(point,nearest_point)}
```
