```python
def part2(filename, debug=False):

    data = read_input(filename)
    maxx = int(max([item[0] for item in data]))+1
    maxy = int(max([item[1] for item in data]))+1
    
    board = [["." for i in range(maxx)] for j in range(maxy)]
    if debug:
        print("blank board")
        for row in board:
            print("".join(row))
    for point in data:
        board[point[1]][point[0]] = "#"
    if debug:
        print("placing #s")
        for row in board:
            print("".join(row))
    for row in board:
        inds = [i for i in range(len(row)) if row[i]=="#"]
        if len(inds) > 1:
            for i in range(inds[0]+1,inds[-1]+1):
                if row[i]==".":
                    row[i] = "X"
    if debug:
        print("filling in rows with Xs")
        for row in board:
            print("".join(row))
    transposed_board = [list(row) for row in zip(*board)]
    if debug:
        print("transposing board")
        for row in transposed_board:
            print("".join(row))
    for row in transposed_board:
        inds = [i for i in range(len(row)) if row[i] in ["#","X"]]
        if len(inds) > 1:
            for i in range(inds[0]+1,inds[-1]+1):
                if row[i]==".":
                    row[i] = "X"
    if debug:
        print("filling in columns with Xs")
        for row in transposed_board:
            print("".join(row))
    board = [list(row) for row in zip(*transposed_board)]
    if debug:
        print("transposing board back")
        for row in board:
            print("".join(row))
    max_size = 0
    for point1 in data:
        for point2 in data:
            minx = min([point1[0],point2[0]])
            maxx = max([point1[0],point2[0]])
            miny = min([point1[1],point2[1]])
            maxy = max([point1[1],point2[1]])
            if (((board[miny][minx]=="#") and (board[maxy][maxx])) or ((board[miny][maxx]=="#") and (board[maxy][minx]))):
                block = []
                for row in board[miny:maxy+1]:
                    block.append(row[minx:maxx+1])
                if any("." in row for row in block):
                    continue
                else:
                    size = (maxy+1-miny)*(maxx+1-minx)
                    if size > max_size:
                        max_size = size
                    if len(block)>1:
                        if debug:
                            print(f"{max_size=} {size=} {point1=} {point2=} {minx=} {maxx=} {miny=} {maxy=} {len(board)=} {len(board[0])=} {len(block)=} {len(block[0])=}")
                            print(" ","".join([hex(idx)[2:] for idx in range(minx,maxx+1)]))
                            for idx,row in enumerate(block):
                                print(idx+miny,"".join(row))
    return max_size
```

```python
assert part2("test.txt", debug=False) == 24
```
