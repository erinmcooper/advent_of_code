```python
def part1(filename):
    data = read_input(filename)
    
    max_area = 0
    corners = []
    store_temp = []
    for i in range(len(data)):
        for j in range(len(data)):
            if i == j:
                continue
            else:
                temp = abs(data[i] - data[j])
                area = (temp[0]+1)*(temp[1]+1)
                if area > max_area:
                    max_area = area
                    corners = [data[i],data[j]]
                    store_temp = temp
    print(f"{max_area=}")
    return max_area
```

```python
data = read_input("test.txt")
```

```python
maxx = int(max([item[0] for item in data]))
maxy = int(max([item[1] for item in data]))
```

```python
board = [["." for i in range(maxy+2)] for j in range(maxx+2)]
```

```python
board
```

```python
for point in data:
    print(point)
    board[maxx-int(point[0])][maxy-int(point[1])] = "#"
```

```python
for i in range(len(board)-1,0,-1):
    print("".join([board[i][j] for j in range(len(board[i])-1,0,-1)]))
```
