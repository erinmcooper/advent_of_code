```python
def read_input(filename):
    data = []
    with open(filename, "r") as f:
        for line in f.readlines():
            lights = line.split()[0][1:-1]
            lights = tuple([i for i in range(len(lights)) if lights[i]=="#"])
            data.append({
                "lights": lights, 
                "buttons": [tuple([int(i) for i in thing.split("(")[1].split(")")[0].split(",")]) for thing in line.split()[1:-1]],
                "joltages": [int(i) for i in line.split()[-1][1:-1].split(",")]
            })
            
    return data
```

```python
from itertools import combinations
```

```python
from collections import Counter
```

```python
def test_n_buttons(light, buttons, n=2, debug=False):
    for idx, thing in enumerate(combinations(buttons, n)):
        all_elements = [item for t in thing for item in t]
        counts = Counter(all_elements)
        result = [item for item in counts if counts[item] % 2 == 1]
        result.sort()
        result = tuple(result)
        if debug:
            print(n, idx, thing, counts, result, result==key)
        if result == light:
            return True
```

```python
def part1(filename, debug=False):
    
    data = read_input(filename)
    button_presses = 0
    # test 1 button press
    for idx, item in enumerate(data):
        light = item["lights"]
        buttons = item["buttons"]
        if debug:
            print(f"{light=}")
        min_presses = 1
        if light in buttons:
            button_presses+=min_presses
            if debug:
                print(f"{min_presses=} {button_presses=}")
            if debug:
                print(f"{idx=} {light=}")
                print(buttons)
            continue
        else:
            found_solution = False
            while not found_solution:
                min_presses+=1
                found_solution = test_n_buttons(light, buttons, n=min_presses, debug=debug)
                if found_solution:
                    button_presses+=min_presses
                    if debug:
                        print(f"{min_presses=} {button_presses=}")
                    break
    print(f"{button_presses=}")
    return button_presses
```
