```python
def read_input(filename):
    data = {}
    with open(filename, "r") as f:
        for line in f.readlines():
            data[line.split()[0][1:-1]] = [tuple([int(i) for i in thing.split("(")[1].split(")")[0].split(",")]) for thing in line.split()[1:-1]]
    return data
```

```python
data = read_input("test.txt")
```
